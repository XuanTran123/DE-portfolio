```python
import kagglehub

# Find the path of the dataset
path = kagglehub.dataset_download("olistbr/brazilian-ecommerce")

print("Path to dataset files:", path)
```

```python
#Display files
files = dbutils.fs.ls(f"file:{path}")
display(files)
```

```python
#Make directory for the dataset in Databricks
dst_dir = "dbfs:/FileStore/kaggle/brazilian-ecommerce"
dbutils.fs.mkdirs(dst_dir)

# Download each file to Databricks
for f in files:
    src_path = f.path             
    dst_path = f"{dst_dir}/{f.name}" 
    dbutils.fs.cp(src_path, dst_path)

print(f"Copied {len(files)} files into {dst_dir}")
```

```python
#Display directory in Databricks
display(dbutils.fs.ls(dst_dir))
```
